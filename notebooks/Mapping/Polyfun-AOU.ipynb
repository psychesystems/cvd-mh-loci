```python
mkdir -p LD_cache

clusters=(afr amr eas eur sas)
sumstats=(mdd2024_afr mdd2024_eas mdd2024_his mdd2025_eur mdd2024_sas)
ns=(70727 58319 16211 14979 1577200)

for ((i=0 ; i < ${#clusters[@]} ; ++i)); do
    cluster=${clusters[i]}
    sums=${sumstats[i]}
    n=${ns[i]}

    python polyfun/finemapper.py \
        --geno acaf_chr11_61000000-63000000_${cluster} \
        --sumstats ${sums}_hg38_snpvar.tsv \
        --n $n \
        --chr 11 \
        --start 61600000 \
        --end 62000000 \
        --method finemap \
        --finemap-exe finemap_v1.4.1_x86_64/finemap_v1.4.1_x86_64 \
        --max-num-causal 5 \
        --allow-missing \
        --cache-dir LD_cache \
        --out ${sums}_polyfun_finemap_acaf_hg38_chr11_61600000-6200000.tsv
done
```
