# Region reference files

[1000 Genomes phase 3](https://www.cog-genomics.org/plink/2.0/resources#phase3_1kg) from plink2 resources.

```python
from pyprojroot.here import here
import polars as pl
```

Extract specified region for each SuperPop. Use bash magic command and pass directory location as an argument to the script.

```python
%%bash -s {here()}
here=$1

for cluster in AFR AMR EAS EUR SAS; do
    plink2 \
        --pfile ${here}/data/reference/1000G/all_hg38 'vzs' \
        --keep-if "SuperPop = $cluster" \
        --chr 11 \
        --from-bp 61000000 \
        --to-bp 63000000 \
        --max-alleles 2 \
        --make-bed \
        --out ${here}/data/processed/reference/all_hg38_${cluster}_chr11_61000000-63000000
done
```
